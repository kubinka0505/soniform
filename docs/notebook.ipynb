### <font class="markdown-google-sans">lissajous 🪱</font>
<img src="https://raw.githubusercontent.com/kubinka0505/soniform/main/docs/img/lissajous.gif" height=250>


```python
import os
import numpy as np
import soundfile as sf
from IPython.display import Audio, display

script_name = "lissajous"
script_file = f"{script_name}.py"

#@markdown ---
#@markdown #### 🗃️ Files

output_path_sound = "{file_name}/{file_name}_{sample_rate}_{frequency}.wav" #@param {type: "string"}

#@markdown ---
#@markdown #### 🔊 Sound

attack = 0.5 #@param {"type": "number"}
duration = 2 #@param {"type": "number"}
release = 0.5 #@param {"type": "number"}

frequency = 130.8127826502993 #@param {"type": "slider", "min": 1, "max": 20000, "step": 0.001}
sample_rate = 44100 #@param [8000, 16000, 22050, 32000, 44100, 48000, 96000, 192000]

#@markdown ---
#@markdown #### 🖼️ Shape

rotation = -45 #@param {"type": "slider", "min": -360, "max": 360, "step": 1}
starting_point = 0 #@param {"type": "number"}

samples = 4000 #@param {"type": "slider", "min": 100, "max": 10000, "step": 100}
points = 2000 #@param {"type": "slider", "min": 100, "max": 10000, "step": 100}
margin = 0.5 #@param {"type": "slider", "min": 0, "max": 1, "step": 0.01}

#@markdown ---
#@markdown #### 🔄 Switches

reverse = False #@param {type: "boolean"}
show_help = False #@param {type: "boolean"}

#-=-=-=-#

os.chdir(path_repo)

files_populated = populate_uploads(directory_files)

for idx, file_input in enumerate(files_populated):
	file_name = os.path.splitext(os.path.basename(file_input))[0]

	current_output_sound = output_path_sound.format(
		file_name = file_name,
		sample_rate = sample_rate,
		frequency = frequency
	)

	os.makedirs(
		os.path.dirname(current_output_sound),
		exist_ok = True
	)

	flag_reverse = "-rev" if reverse else ""
	flag_help = "--help" if show_help else ""

	#-=-=-=-#

	!python "$script_file" {flag_help} \
		-i "$file_input" \
		-o "$current_output_sound" \
		-a {attack} \
		-d {duration} \
		-r {release} \
		-freq {frequency} \
		-sr {sample_rate} \
		-rot {rotation} \
		-ss {starting_point} \
		-s {samples} \
		-p {points} \
		-m {margin} \
		{flag_reverse}

	#-=-=-=-#

	if os.path.exists(current_output_sound):
		print(f"Created: {current_output_sound}")

		display(
			Audio(
				filename = current_output_sound,
				autoplay = True
			)
		)
	else:
		print(f"Output not found: {current_output_sound}")

	#-=-=-=-#

	if idx != len(files_populated) - 1:
		print()
```

### <font class="markdown-google-sans">specimg 🖼️</font>
<img src="https://raw.githubusercontent.com/kubinka0505/soniform/main/docs/img/specimg.png" width=100>

```python
import os
from IPython.display import Audio, display

script_name = "specimg"
script_file = f"{script_name}.py"

#@markdown ---
#@markdown #### 🗃️ Files

output_path_sound = "{file_name}/{file_name}_{sample_rate}_{frequency_min}-{frequency_max}.wav" #@param {type: "string"}

#@markdown ---
#@markdown #### 🎛️ Spectrogram Processing

frequency_min = 15000 #@param {"type": "number"}
frequency_max = 16000 #@param {"type": "number"}

target_peak_db = -1 #@param {"type": "number"}
min_level_db = None #@param {type: "raw"}

griffin_lim = 0 #@param {"type": "slider", "min": 0, "max": 200, "step": 1}

frequency_mode = "linear" #@param ["linear", "log", "exp"]
phase_mode = "random" #@param ["random", "zero"]

#@markdown ---
#@markdown #### 🔊 Sound

sample_rate = 44100 #@param [8000, 16000, 22050, 32000, 44100, 48000, 96000, 192000]
duration = 1 #@param {"type": "number"}

n_fft = 2048 #@param {"type": "slider", "min": 256, "max": 8192, "step": 256}
hop_ratio = 0 #@param {"type": "number"}
window = "hann" #@param ["hann", "hamming", "blackman"]

fade_ms = 0 #@param {"type": "number"}

#@markdown ---
#@markdown #### 🖼️ Image Processing

gamma = 1 #@param {"type": "number"}
contrast = 1 #@param {"type": "number"}
brightness = 1 #@param {"type": "number"}

threshold = None #@param {type: "raw"}
vector_render_scale = 32 #@param {"type": "number"}

#@markdown ---
#@markdown #### 🔄 Switches

invert = False #@param {type: "boolean"}
flip_horizontal = False #@param {type: "boolean"}
flip_vertical = False #@param {type: "boolean"}

#@markdown ---
show_help = False #@param {type: "boolean"}

#-=-=-=-#

os.chdir(path_repo)

files_populated = populate_uploads(directory_files)

for idx, file_input in enumerate(files_populated):

	file_name = os.path.splitext(os.path.basename(file_input))[0]

	current_output_sound = os.path.join(
		"outputs",
		script_name,
		output_path_sound.format(
			file_name = file_name,
			sample_rate = sample_rate,
			frequency_min = frequency_min,
			frequency_max = frequency_max
		)
	)

	os.makedirs(
		os.path.dirname(current_output_sound),
		exist_ok = True
	)

	print(current_output_sound)

	flag_inv = "-inv" if invert else ""
	flag_fx = "-fx" if flip_horizontal else ""
	flag_fy = "-fy" if flip_vertical else ""

	flag_help = "--help" if show_help else ""

	min_level_arg = "" if min_level_db is None else f"-ml {min_level_db}"
	threshold_arg = "" if threshold is None else f"-vt {threshold}"

	#-=-=-=-#

	!python "$script_file" {flag_help} \
		-i "$file_input" \
		-o "{current_output_sound}" \
		-fmin {frequency_min} \
		-fmax {frequency_max} \
		-peak {target_peak_db} \
		{min_level_arg} \
		-g {griffin_lim} \
		-fm {frequency_mode} \
		-pm {phase_mode} \
		-sr {sample_rate} \
		-d {duration} \
		-fft {n_fft} \
		-hl {hop_ratio} \
		-w {window} \
		-fms {fade_ms} \
		-ig {gamma} \
		-ic {contrast} \
		-ib {brightness} \
		{threshold_arg} \
		-vs {vector_render_scale} \
		{flag_inv} \
		{flag_fx} \
		{flag_fy}

	#-=-=-=-#

	if os.path.exists(current_output_sound) and not show_help:
		display(
			Audio(
				filename = current_output_sound,
				autoplay = True
			)
		)
	else:
		print(f"Output not found: {current_output_sound}")

	#-=-=-=-#

	if idx != len(files_populated) - 1:
		print()
```

### <font class="markdown-google-sans">wavegraph 📊</font>
<img src="https://raw.githubusercontent.com/kubinka0505/soniform/main/docs/img/wavegraph.gif" height=125>

```python
import os
from IPython.display import Image

script_name = "wavegraph"
script_file = f"{script_name}.py"

#@markdown ---
#@markdown #### <font class="markdown-google-sans">🗃️ Files</font>

output_path_image = "{file_name}/{file_name}_{samples}.gif" #@param {type: "string"}
output_path_sound = "{file_name}/{file_name}_{sample_rate}_{frequency}.wav" #@param {type: "string"}

#@markdown ---
#@markdown #### <font class="markdown-google-sans">🖼️ Image</font>

color = "" #@param {type: "string"}
frames = 300 # @param {"type": "slider", "min": 2, "max": 3000, "step": 1}
framerate = 50 #@param {"type": "slider", "min": 1, "max": 50, "step": 1}
samples = 1500 # @param {"type":"slider", "min": 100, "max": 3000, "step": 1}
theta_resolution = 2048 #@param {"type": "slider", min: 128, "max": 4096, "step": 16}
dpi = 220 #@param {"type": "slider", "min": 25, "max": 400, "step": 1}

#@markdown ---
#@markdown #### <font class="markdown-google-sans">🔊 Sound</font>

frequency = 130.8127826502993 #@param {"type": "slider", "min": 1, "max": 20000, "step": 0.001}
phase = 0 #@param {"type": "slider", "min": 0, "max": 360, "step": 1}
sample_rate = 44100 #@param [8000, 16000, 20000, 22050, 24000, 32000, 40000, 44100,  48000, 88200, 96000, 192000]

#@markdown ---
#@markdown #### 🔄 Switches

crop = True #@param {type: "boolean"}
trace_full_wave = False #@param {type: "boolean"}

#@markdown ---

show_help = False #@param {type: "boolean"}

#-=-=-=-#

flag_no_crop = not crop
flag_trace = not trace_full_wave

flag_help = "--help" if show_help else ""

#-=-=-=-#

os.chdir(path_repo)

files_populated = populate_uploads(directory_files)

for idx, file_input in enumerate(files_populated):
	file_name = os.path.splitext(os.path.basename(file_input))[0]

	output_path_image = output_path_image.format(
		prog_name = script_name,
		file_name = file_name,
		samples = samples,
		frequency = frequency,
		phase = phase,
		sample_rate = sample_rate,
		dpi = dpi,
		color = color,
		frames = frames,
		framerate = framerate,
		theta_resolution = theta_resolution,
	)

	#-=-=-=-#

	!python "$script_file" {flag_help} \
		-i "$file_input" \
		-oi "{output_path_image}" \
		-os "{output_path_sound}" \
		-c "{color}" \
		-f {frames} \
		-fps {framerate} \
		-s {samples} \
		-tr {theta_resolution} \
		-dpi {dpi} \
		-freq {frequency} \
		-p {phase} \
		-sr {sample_rate} \
		-nc {flag_no_crop} \
		-nt {flag_trace}

	#-=-=-=-#

	image_path = os.path.join(path_repo, output_path_image)

	if os.path.exists(image_path) and not show_help:
		print()
		display(Image(filename = image_path, height = 150))
		print()
	else:
		print(f"Image not found at {image_path}")

	#-=-=-=-#

	if idx != len(files_populated) - 1:
		print()
```

## 🗃️ <font class="markdown-google-sans">Download archive with `outputs`</font>

```python
import os
import zipfile
import datetime
from google.colab import files

#-=-=-=-#

directory_outputs = os.path.join(path_repo, "outputs")

archive = os.path.join(
	"/content",
	"_".join((
	name_repo,
		*datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S").split()
	))
)
archive = archive.replace(" ", "_") + ".zip"

#-=-=-=-#
# Creating archive

print("Writing files...")

to_write = [str(file.resolve()) for file in Path(directory_outputs).rglob("*") if file.is_file()]
to_write.sort(key = os.path.getsize)
to_write.sort(key = len)
to_write.sort(key = str)

total = len(to_write)

with zipfile.ZipFile(archive, "w", compression = zipfile.ZIP_LZMA, allowZip64 = True, compresslevel = 9) as arc:
	current = 0

	for file in to_write:
		current += 1
		current_fmt = str(current).rjust(len(str(total)), "0")

		arcname = os.path.relpath(file, directory_outputs)

		print(f'\t[{current_fmt}/{total}] "{arcname}"')

		arc.write(file, arcname)

#-=-=-=-#

print("\n" + "—" * 32 + "\n")
print("Done!")

files.download(archive)
```
